# 02 · Live training (MA-4)

Trains one model with the live dashboard drawn in the notebook. Run from `cnn/notebooks/`; paths below are relative to it.

For comparing many runs, use TensorBoard instead: `tensorboard --logdir outputs` from `cnn/`.

```python
%matplotlib inline
from nbacnn.config import load_config
from nbacnn.train import train
from nbacnn.live_plot import LivePlot

cfg = load_config("../configs/default.yaml",
                  ["data_dir=../../data/processed", "out_dir=../outputs/runs", "run_name=notebook"])
c = cfg["convergence"]
```

```python
live = LivePlot(c["target_top1"], c["too_good_top1"], mode="notebook")
run_dir = train(cfg, live=live)
run_dir
```

```python
import json, pandas as pd
print(json.dumps(json.load(open(run_dir / "summary.json")), indent=2))
pd.read_csv(run_dir / "metrics.csv").tail()
```
