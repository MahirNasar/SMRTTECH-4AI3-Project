```python
import matplotlib.pyplot as plt
from nbacnn.data import load_split

# 1. Load the val split
val_data = load_split("val", data_dir="../../data/processed")

# 2. Find Paul Pierce's index
idx = val_data.ids.index("piercpa01")

# 3. Plot his career grid
plt.figure(figsize=(10, 5))
# grid shape is (P, 23, 26); converting to numpy for matplotlib
plt.imshow(val_data.grid[idx].numpy(), aspect="auto")
plt.title("Paul Pierce (piercpa01) - Career Grid")
plt.xlabel("Age Index (0 = Age 18)")
plt.ylabel("Channels (23 stats/positions)")
plt.colorbar()
plt.show()

# 4. Plot his two shot chart channels
# chart shape is (P, 2, 24, 25)
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Channel 0: freq_rel
im0 = axes[0].imshow(val_data.chart[idx, 0].numpy(), aspect="auto")
axes[0].set_title("Shot Chart: Relative Frequency")
fig.colorbar(im0, ax=axes[0])

# Channel 1: eff_rel
im1 = axes[1].imshow(val_data.chart[idx, 1].numpy(), aspect="auto")
axes[1].set_title("Shot Chart: Relative Efficiency")
fig.colorbar(im1, ax=axes[1])

plt.show()
```
